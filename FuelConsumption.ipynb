## CO₂ Emissions Distribution Analysis

### Visualization
A histogram with a Kernel Density Estimate (KDE) curve was plotted to examine the distribution of the `CO2EMISSIONS` variable.

```python
sns.histplot(Fuel['CO2EMISSIONS'], kde=True)
plt.show()
```

### Insights

1. The histogram shows how CO₂ emission values are distributed across different vehicles.

2. The peak of the histogram represents the most common emission range among vehicles in the dataset.

3. If the distribution is right-skewed (long tail toward higher values), it indicates that most vehicles produce lower to moderate emissions, while a smaller number of vehicles produce very high emissions.

4. The KDE curve helps visualize the overall shape of the distribution and makes it easier to identify patterns compared to bars alone.

5. Any isolated bars far from the main distribution may indicate potential outliers or vehicles with unusually high CO₂ emissions.

6. The spread of the histogram indicates the variability in vehicle emissions. A wider spread suggests greater differences in emissions between vehicles.

### Conclusion

The distribution of CO₂ emissions provides an initial understanding of emission patterns in the dataset and helps identify common emission levels, variability, and potential outliers for further investigation.

## Interpretation

The CO₂ emissions distribution is positively skewed. Most vehicles have emissions concentrated in the lower to moderate range, while a smaller number of vehicles exhibit significantly higher emission levels. This suggests that high-emission vehicles are less common but contribute substantially to overall emissions. The presence of a long right tail indicates potential outliers, which may correspond to larger engines or vehicles with higher fuel consumption.

## Correlation Analysis of Fuel Consumption Features

### Visualization

```python
corr = Fuel[num_cols].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

### Insights

1. The correlation heatmap shows the strength and direction of the relationship between numerical variables in the dataset.

2. Values closer to **+1** indicate a strong positive correlation, meaning that as one variable increases, the other tends to increase as well.

3. Values closer to **-1** indicate a strong negative correlation, meaning that as one variable increases, the other tends to decrease.

4. Values near **0** suggest little or no linear relationship between the variables.

5. The heatmap typically shows a strong positive correlation between:
   - `ENGINESIZE` and `CO2EMISSIONS`
   - `CYLINDERS` and `CO2EMISSIONS`
   - Fuel consumption measures (`FUELCONSUMPTION_CITY`, `FUELCONSUMPTION_HWY`, `FUELCONSUMPTION_COMB`) and `CO2EMISSIONS`

6. `FUELCONSUMPTION_COMB` generally exhibits the strongest correlation with `CO2EMISSIONS`, indicating that combined fuel consumption is one of the best predictors of vehicle emissions.

7. The strong positive correlation between `ENGINESIZE` and `CYLINDERS` suggests that vehicles with larger engines generally have more cylinders.

8. `FUELCONSUMPTION_COMB_MPG` often shows a strong negative correlation with `CO2EMISSIONS`. This indicates that vehicles with higher fuel efficiency (more miles per gallon) tend to produce lower carbon emissions.

### Conclusion

The heatmap reveals that engine size, number of cylinders, and fuel consumption are major factors influencing CO₂ emissions. Vehicles with larger engines and higher fuel consumption generally emit more CO₂, while vehicles with better fuel efficiency tend to emit less. These variables can be considered important predictors for building a CO₂ emission prediction model.

## CO₂ Emissions Distribution and Outlier Analysis

### Visualization

```python
sns.boxplot(y=Fuel['CO2EMISSIONS'])
plt.show()
```

### Insights

1. The box plot provides a summary of the distribution of CO₂ emissions, including the median, spread, and potential outliers.

2. The horizontal line inside the box represents the median CO₂ emission value, indicating the typical emission level of vehicles in the dataset.

3. The box represents the interquartile range (IQR), which contains the middle 50% of emission values. A larger box indicates greater variability in emissions.

4. The whiskers show the range of typical values, excluding extreme observations.

5. Any points beyond the whiskers are considered potential outliers. These vehicles have unusually high or low CO₂ emissions compared to the majority of vehicles.

6. If more outliers appear above the upper whisker than below the lower whisker, this suggests that some vehicles produce exceptionally high emissions.

7. The presence of outliers indicates that certain vehicles differ significantly from the rest of the dataset, likely due to factors such as larger engine sizes, higher cylinder counts, or increased fuel consumption.

### Conclusion

The box plot reveals the overall spread of CO₂ emissions and highlights potential outliers. The presence of high-emission outliers suggests that a small number of vehicles contribute disproportionately to carbon emissions, making them important candidates for further investigation.

### Observation

The box plot shows the distribution of CO₂ emissions across all vehicles. Most emission values are concentrated within a specific range, while several observations lie outside the whiskers, indicating potential outliers.

### Insight

The existence of outliers suggests that some vehicles emit significantly more CO₂ than the average vehicle. These extreme values may be associated with larger engines, more cylinders, or higher fuel consumption rates. Such vehicles can have a substantial impact on overall emission levels.

# Sustainability

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Style
sns.set_theme(style="whitegrid")

# Numerical columns
num_cols = [
    'ENGINESIZE',
    'CYLINDERS',
    'FUELCONSUMPTION_CITY',
    'FUELCONSUMPTION_HWY',
    'FUELCONSUMPTION_COMB',
    'FUELCONSUMPTION_COMB_MPG',
    'CO2EMISSIONS'
]

# Create figure
fig, axes = plt.subplots(2, 3, figsize=(18, 12))

# 1. CO2 Distribution
sns.histplot(
    Fuel['CO2EMISSIONS'],
    kde=True,
    color='royalblue',
    ax=axes[0, 0]
)
axes[0, 0].set_title('CO2 Emissions Distribution')

# 2. Engine Size Distribution
sns.histplot(
    Fuel['ENGINESIZE'],
    kde=True,
    color='green',
    ax=axes[0, 1]
)
axes[0, 1].set_title('Engine Size Distribution')

# 3. Engine Size vs CO2
sns.scatterplot(
    x='ENGINESIZE',
    y='CO2EMISSIONS',
    data=Fuel,
    color='crimson',
    ax=axes[0, 2]
)
axes[0, 2].set_title('Engine Size vs CO2 Emissions')

# 4. Fuel Type vs CO2
sns.barplot(
    x='FUELTYPE',
    y='CO2EMISSIONS',
    hue='FUELTYPE',
    data=Fuel,
    palette='viridis',
    legend=False,
    ax=axes[1, 0]
)
axes[1, 0].set_title('Average CO2 Emissions by Fuel Type')

# 5. Correlation Heatmap
corr = Fuel[num_cols].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    ax=axes[1, 1]
)
axes[1, 1].set_title('Correlation Heatmap')

# 6. Boxplot of CO2
sns.boxplot(
    y=Fuel['CO2EMISSIONS'],
    color='orange',
    ax=axes[1, 2]
)
axes[1, 2].set_title('CO2 Emissions Outliers')

plt.tight_layout()
plt.show()
````
